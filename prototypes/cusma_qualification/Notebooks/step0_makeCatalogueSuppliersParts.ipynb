```python
"""STEP 0: write the input csv for step 2 ->  Catalogue/boots.csv, parts.csv, suppliers.csv

This creates the boots catalog, supplier list and boot parts. In this step, boots are hand typed.
the fixed suppliers, the 11 catalogue boots, and the parts in each boot
The Word "Boot Catalogue" is a readable view of these same files.

To change a boot after sign-off: edit it HERE, re-run this cell, then re-run steps 2 and 3.

Sources: the 7 styles come from the workbook's Archetypes tab; the 3 Glacier boots are the
workbook's Example BOM A, B and C; tariff codes come from Nazia's Verified_Component_Library;
countries from its "origin options"; costs, prices and supplier names are team estimates
(all fictional).
"""

from google.colab import drive
drive.mount('/content/drive', force_remount=True)

import csv, pathlib

# ---- FOLDERS ----
DRIVE            = pathlib.Path('/content/drive/MyDrive/woodb-agent-project/prototypes/cusma_qualification')
CATALOGUE_FOLDER = DRIVE / 'Catalogue'                            # writes boots, parts, suppliers here
WORKBOOK         = DRIVE / 'Rules' / 'Nazia_Rules_Workbook.xlsx'  # reads Nazia's codes from here

# ---------------------------------------------------------------------------------------
# 1. FIXED SUPPLIERS: (id, name, country, what they supply, declaration status, expiry, language)
#    Declarations matter only for Canada (CA), the US and Mexico (MX): a part from there counts
#    as North American only with a valid, unexpired declaration (Nazia, Assumption #2).
# ---------------------------------------------------------------------------------------
SUPPLIERS = [
    ('SUP-010', 'Hartwig Polymere GmbH', 'DE', 'TPU pellets', 'not_applicable', '', 'en'),
    ('SUP-011', 'Brightline Color Inc.', 'US', 'Colour masterbatch', 'valid', '2027-06-30', 'en'),
    ('SUP-012', 'Prairie Foam Systems LLC', 'US', 'PU foam system', 'valid', '2027-03-31', 'en'),
    ('SUP-013', 'Jade River Textiles Co.', 'CN', 'Knit lining; trims', 'not_applicable', '', 'en'),
    ('SUP-014', 'Siam Latex Trading Ltd.', 'TH', 'Natural rubber', 'not_applicable', '', 'en'),
    ('SUP-015', 'Gulfport Elastomers Inc.', 'US', 'SBR/BR rubber', 'valid', '2027-01-31', 'en'),
    ('SUP-016', 'Chimie Laurentide inc.', 'CA', 'Carbon black and cure chemicals', 'valid', '2027-05-31', 'fr'),
    ('SUP-017', 'Ningbo Safeguard Parts Co.', 'CN', 'Composite toe caps', 'not_applicable', '', 'en'),
    ('SUP-018', 'Aramid Technik GmbH', 'DE', 'Composite puncture plates', 'not_applicable', '', 'en'),
    ('SUP-019', 'Plantillas del Pacifico S.A. de C.V.', 'MX', 'Insoles', 'valid', '2026-12-31', 'es'),
    ('SUP-020', 'Gel-Confort Quebec inc.', 'CA', 'Gel midsoles', 'valid', '2027-02-28', 'fr'),
    ('SUP-022', 'Cartonnage Mauricie inc.', 'CA', 'Retail boxes', 'valid', '2027-04-30', 'fr'),
    ('SUP-023', 'Mekong Sole Works Co.', 'VN', 'Finished outsoles', 'not_applicable', '', 'en'),
    ('SUP-024', 'Dongguan Upper Craft Co.', 'CN', 'Pre-formed uppers', 'not_applicable', '', 'en'),
    ('SUP-025', 'Keystone Safety Components LLC', 'US', 'Composite toe caps', 'valid', '2027-08-31', 'en'),
    ('SUP-026', 'Conceria Valdarno S.r.l.', 'IT', 'Finished bovine leather', 'not_applicable', '', 'en'),
    ('SUP-027', 'Hangzhou Weave Tech Co.', 'CN', 'Nylon upper fabric', 'not_applicable', '', 'en'),
    ('SUP-028', 'Hebei Felt Products Co.', 'CN', 'Wool-blend felt sheet', 'not_applicable', '', 'en'),
    ('SUP-029', 'Foshan Hardware Fittings Co.', 'CN', 'Plastic buckles, eyelets, laces', 'not_applicable', '', 'en'),
    ('SUP-030', 'Shenzhen NeoFlex Materials Co.', 'CN', 'Jersey-laminated neoprene sheet', 'not_applicable', '', 'en'),
    ('SUP-031', 'Tianjin Steelcap Co.', 'CN', 'Steel toe caps and plates', 'not_applicable', '', 'en'),
    ('SUP-032', 'Xiamen Comfort Insole Co.', 'CN', 'Insoles', 'not_applicable', '', 'en'),
    ('SUP-033', 'Suole Alpine S.p.A.', 'IT', 'Premium branded rubber outsoles', 'not_applicable', '', 'en'),
    ('SUP-034', 'Bay State Outsole Co.', 'US', 'Premium branded rubber outsoles (US plant)', 'valid', '2027-09-30', 'en'),
]

# ---------------------------------------------------------------------------------------
# 2. BOOTS: one entry per catalogue boot. "variant_of" links a deliberate variant to its base.
# ---------------------------------------------------------------------------------------
BOOT_FIELDS = ['sku', 'style_no', 'style_name', 'variant_of', 'archetype', 'construction', 'upper_material', 'upper_class', 'upper_main_share', 'gender', 'height_in', 'covers_knee', 'toe_type', 'plate_type', 'liner', 'outsole_source', 'labour_cad', 'overhead_cad', 'price_usd', 'tests']
BOOTS = [
    dict(sku='NB-0101', style_no='S1', style_name='Classic Knee', variant_of='', archetype='All-rubber knee boot', construction='moulded', upper_material='Natural rubber', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='18.0', covers_knee='yes', toe_type='none', plate_type='none', liner='knit lining', outsole_source='in-house', labour_cad='13.00', overhead_cad='8.50', price_usd='58.00', tests='Baseline: comfortably qualifies'),
    dict(sku='NB-0102', style_no='S2', style_name='Steelguard', variant_of='', archetype='Steel-toe rubber safety boot', construction='moulded', upper_material='Natural rubber', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='15.0', covers_knee='no', toe_type='steel', plate_type='steel', liner='knit lining', outsole_source='in-house', labour_cad='13.50', overhead_cad='9.00', price_usd='72.00', tests='Metal toe decides the code (6401.10); imported toe cap saved by the 10% allowance'),
    dict(sku='NB-0412', style_no='S3', style_name='Glacier Pro', variant_of='', archetype='Insulated TPU safety boot', construction='moulded', upper_material='TPU moulded', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='17.5', covers_knee='yes', toe_type='composite', plate_type='composite', liner='PU insulation + knit lining', outsole_source='in-house', labour_cad='14.00', overhead_cad='9.50', price_usd='64.00', tests='Composite toe does not trigger 6401.10; toe cap saved by the 10% allowance'),
    dict(sku='NB-0413', style_no='S3', style_name='Glacier Pro V', variant_of='NB-0412', archetype='Insulated TPU safety boot', construction='moulded', upper_material='TPU moulded', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='17.5', covers_knee='yes', toe_type='composite', plate_type='composite', liner='PU insulation + knit lining', outsole_source='bought', labour_cad='11.50', overhead_cad='8.50', price_usd='64.00', tests='Bought outsoles push RVC to the edge: goes to a person'),
    dict(sku='NB-0414', style_no='S3', style_name='Glacier Lite', variant_of='NB-0412', archetype='Low-cost import-heavy variant', construction='moulded', upper_material='Pre-formed TPU upper (bought)', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='17.5', covers_knee='yes', toe_type='composite', plate_type='composite', liner='PU insulation + knit lining', outsole_source='in-house', labour_cad='11.00', overhead_cad='8.00', price_usd='52.00', tests='Bought upper breaks the rule even though RVC is above 55%'),
    dict(sku='NB-0104', style_no='S4', style_name='Garden Ankle', variant_of='', archetype='Ankle rubber boot', construction='moulded', upper_material='Natural rubber', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='6.0', covers_knee='no', toe_type='none', plate_type='none', liner='knit lining', outsole_source='in-house', labour_cad='6.50', overhead_cad='4.50', price_usd='32.00', tests='Cheap boot, high US duty if it fails: biggest saving per dollar'),
    dict(sku='NB-0105', style_no='S5', style_name='Tundra Pac', variant_of='', archetype='Pac / winter boot', construction='stitched', upper_material='Nylon upper on rubber bottom', upper_class='textile', upper_main_share='0.65', gender='unisex', height_in='12.0', covers_knee='no', toe_type='none', plate_type='none', liner='felt liner sewn in-house', outsole_source='in-house', labour_cad='14.50', overhead_cad='9.50', price_usd='95.00', tests='Upper material decides the heading (6404); liner made in-house'),
    dict(sku='NB-0106', style_no='S5', style_name="Tundra Pac Leather (women's)", variant_of='NB-0105', archetype='Pac / winter boot', construction='stitched', upper_material='Leather upper on rubber bottom', upper_class='leather', upper_main_share='0.65', gender='women', height_in='12.0', covers_knee='no', toe_type='none', plate_type='none', liner='felt liner sewn in-house', outsole_source='in-house', labour_cad='14.50', overhead_cad='9.50', price_usd='120.00', tests='Qualifies, but the 50% Section 338 tariff applies anyway'),
    dict(sku='NB-0107', style_no='S6', style_name='Marsh Neo', variant_of='', archetype='Neoprene-shaft boot', construction='moulded foot, stitched shaft', upper_material='Jersey-laminated neoprene shaft', upper_class='neoprene', upper_main_share='0.80', gender='unisex', height_in='16.0', covers_knee='yes', toe_type='none', plate_type='none', liner='neoprene (part of shaft)', outsole_source='in-house', labour_cad='12.50', overhead_cad='8.50', price_usd='78.00', tests='Origin is fine, but the code needs a person: rubber or textile upper?'),
    dict(sku='NB-0109', style_no='S6', style_name='Marsh Neo Pro', variant_of='NB-0107', archetype='Neoprene-shaft boot', construction='moulded foot, stitched shaft', upper_material='Jersey-laminated neoprene shaft', upper_class='neoprene', upper_main_share='0.80', gender='unisex', height_in='16.0', covers_knee='yes', toe_type='none', plate_type='none', liner='neoprene (part of shaft)', outsole_source='bought', labour_cad='11.50', overhead_cad='8.00', price_usd='110.00', tests='Premium imported sole pulls value close to 55%: goes to a person (and the code is still a judgment call)'),
    dict(sku='NB-0108', style_no='S7', style_name='Value Knee', variant_of='', archetype='Low-cost import-heavy variant', construction='assembled', upper_material='Pre-formed rubber upper (bought)', upper_class='rubber_plastic', upper_main_share='1.00', gender='unisex', height_in='16.0', covers_knee='yes', toe_type='steel', plate_type='none', liner='included in bought upper', outsole_source='bought', labour_cad='8.00', overhead_cad='6.00', price_usd='39.00', tests='Fails both tests: bought upper and low RVC'),
]

# ---------------------------------------------------------------------------------------
# 3. PARTS in each boot: (component, description, tariff code, supplier id, cost CAD per pair)
#    The country of origin is taken from the supplier. 6406.10 = upper or part of an upper
#    (incl. toe caps): an EXCLUDED part if imported.
# ---------------------------------------------------------------------------------------
PARTS = {
    'NB-0101': [   # Classic Knee
        ('Upper/shell', 'Hand-laid natural rubber shell (from in-house compound)', '4001.22', 'SUP-014', '4.20'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.50'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '1.30'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '1.40'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.20'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.35'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '0.95'),
    ],
    'NB-0102': [   # Steelguard
        ('Outsole/shell compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '4.40'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.60'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '1.30'),
        ('Safety toe', 'Steel toe cap', '6406.10', 'SUP-031', '2.20'),
        ('Puncture plate', 'Steel midsole plate', '6406.90', 'SUP-031', '2.00'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '1.40'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.30'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.40'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.00'),
    ],
    'NB-0412': [   # Glacier Pro
        ('Upper', 'TPU compound pellets, moulded in-house', '3909.50', 'SUP-010', '7.40'),
        ('Colourant', 'Colour masterbatch', '3206.49', 'SUP-011', '0.35'),
        ('Insulation', 'PU foam system, foamed in place', '3907.29 / 3909.31', 'SUP-012', '3.10'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '1.60'),
        ('Outsole compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '1.90'),
        ('Outsole compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.20'),
        ('Outsole compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '1.10'),
        ('Safety toe', 'Composite toe cap', '6406.10', 'SUP-017', '2.80'),
        ('Puncture plate', 'Composite aramid plate', '6406.90', 'SUP-018', '3.40'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.50'),
        ('Midsole', 'Gel midsole', '6406.90', 'SUP-020', '1.20'),
        ('Trims', 'Labels, safety patch, pull loop', '5807.10', 'SUP-013', '0.45'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.05'),
    ],
    'NB-0413': [   # Glacier Pro V
        ('Upper', 'TPU compound pellets, moulded in-house', '3909.50', 'SUP-010', '7.40'),
        ('Colourant', 'Colour masterbatch', '3206.49', 'SUP-011', '0.35'),
        ('Insulation', 'PU foam system, foamed in place', '3907.29 / 3909.31', 'SUP-012', '3.10'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '1.60'),
        ('Outsole', 'Finished rubber outsole, bought', '6406.20', 'SUP-023', '6.50'),
        ('Safety toe', 'Composite toe cap', '6406.10', 'SUP-017', '2.80'),
        ('Puncture plate', 'Composite aramid plate', '6406.90', 'SUP-018', '3.40'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.50'),
        ('Midsole', 'Gel midsole', '6406.90', 'SUP-020', '1.20'),
        ('Trims', 'Labels, safety patch, pull loop', '5807.10', 'SUP-013', '0.45'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.05'),
    ],
    'NB-0414': [   # Glacier Lite
        ('Upper', 'Pre-formed TPU upper, bought', '6406.10', 'SUP-024', '11.00'),
        ('Colourant', 'Colour masterbatch', '3206.49', 'SUP-011', '0.35'),
        ('Insulation', 'PU foam system, foamed in place', '3907.29 / 3909.31', 'SUP-012', '3.10'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '1.60'),
        ('Outsole compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '1.90'),
        ('Outsole compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.20'),
        ('Outsole compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '1.10'),
        ('Safety toe', 'Composite toe cap', '6406.10', 'SUP-017', '2.80'),
        ('Puncture plate', 'Composite aramid plate', '6406.90', 'SUP-018', '3.40'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.50'),
        ('Midsole', 'Gel midsole', '6406.90', 'SUP-020', '1.20'),
        ('Trims', 'Labels, safety patch, pull loop', '5807.10', 'SUP-013', '0.45'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.05'),
    ],
    'NB-0104': [   # Garden Ankle
        ('Outsole/shell compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '2.10'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '0.80'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '0.70'),
        ('Lining', 'Polyester knit lining', '6006.32', 'SUP-013', '0.80'),
        ('Insole', 'PU insole', '6406.90', 'SUP-032', '0.60'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.25'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '0.70'),
    ],
    'NB-0105': [   # Tundra Pac
        ('Outsole/shell compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '2.60'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.00'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '0.90'),
        ('Upper', 'Nylon upper fabric', '5407.42', 'SUP-027', '3.20'),
        ('Liner', 'Wool-blend felt sheet, liner sewn in-house', '5602.10', 'SUP-028', '2.40'),
        ('Fittings', 'Plastic buckles, eyelets, laces', '3926.90', 'SUP-029', '1.10'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.10'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.10'),
    ],
    'NB-0106': [   # Tundra Pac Leather (women's)
        ('Outsole/shell compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '2.60'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '1.00'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '0.90'),
        ('Upper', 'Finished bovine leather', '4107.92', 'SUP-026', '5.80'),
        ('Liner', 'Wool-blend felt sheet, liner sewn in-house', '5602.10', 'SUP-028', '2.40'),
        ('Fittings', 'Plastic buckles, eyelets, laces', '3926.90', 'SUP-029', '1.10'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.10'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.10'),
    ],
    'NB-0107': [   # Marsh Neo
        ('Outsole/shell compound', 'Natural rubber TSR 20', '4001.22', 'SUP-014', '2.40'),
        ('Outsole/shell compound', 'SBR / BR synthetic rubber', '4002.19', 'SUP-015', '0.90'),
        ('Outsole/shell compound', 'Carbon black, silica, cure chemicals', '2803.00', 'SUP-016', '0.80'),
        ('Shaft', 'Jersey-laminated neoprene sheet, cut and sewn in-house', '5906.99', 'SUP-030', '7.50'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.20'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.40'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.00'),
    ],
    'NB-0109': [   # Marsh Neo Pro
        ('Outsole', 'Premium branded rubber outsole, bought finished', '6406.20', 'SUP-033', '9.00'),
        ('Shaft', 'Jersey-laminated neoprene sheet, cut and sewn in-house', '5906.99', 'SUP-030', '7.50'),
        ('Insole', 'Anti-fatigue PU insole', '6406.90', 'SUP-019', '1.20'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.40'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '1.00'),
    ],
    'NB-0108': [   # Value Knee
        ('Upper', 'Pre-formed rubber upper with lining, bought', '6406.10', 'SUP-024', '10.50'),
        ('Outsole', 'Finished rubber outsole, bought', '6406.20', 'SUP-023', '5.50'),
        ('Safety toe', 'Steel toe cap', '6406.10', 'SUP-031', '2.00'),
        ('Insole', 'PU insole', '6406.90', 'SUP-032', '0.90'),
        ('Trims', 'Labels, patch, pull loop', '5807.10', 'SUP-013', '0.30'),
        ('Packaging', 'Retail box and tissue', '4819.10', 'SUP-022', '0.90'),
    ],
}
# ---------------------------------------------------------------------------------------
# 4. LINK TO NAZIA'S WORKBOOK: which row of her Verified_Component_Library each part is.
#    Every run, each part's code is checked against her verified code (column J):
#      - her cell has ONE code  -> that code is used (so her updates flow through)
#      - her cell lists several -> our code must be one of them (or inside a 4-digit heading)
#      - part not in her library -> kept, and flagged for her to add
# ---------------------------------------------------------------------------------------

LIBRARY_ROW = {
    'TPU compound pellets, moulded in-house': ('Upper (shaft)', 'TPU moulded'),
    'Hand-laid natural rubber shell (from in-house compound)': ('Upper (shaft)', 'Natural rubber'),
    'Jersey-laminated neoprene sheet, cut and sewn in-house': ('Upper (shaft)', 'Textile-faced neoprene'),
    'Nylon upper fabric': ('Upper (shaft)', 'Nylon or leather above ankle'),
    'Finished bovine leather': ('Upper (shaft)', 'Nylon or leather above ankle'),
    'Pre-formed TPU upper, bought': ('Upper (shaft)', 'Pre-formed upper (bought)'),
    'Pre-formed rubber upper with lining, bought': ('Upper (shaft)', 'Pre-formed upper (bought)'),
    'Natural rubber TSR 20': ('Outsole', 'Rubber, in-house'),
    'SBR / BR synthetic rubber': ('Outsole', 'Rubber, in-house'),
    'Carbon black, silica, cure chemicals': ('Outsole', 'Rubber, in-house'),
    'Finished rubber outsole, bought': ('Outsole', 'Finished outsole (bought)'),
    'Premium branded rubber outsole, bought finished': ('Outsole', 'Finished outsole (bought)'),
    'Gel midsole': ('Midsole', 'Anti-fatigue PU/gel midsole'),
    'Anti-fatigue PU insole': ('Insole', 'Anti-fatigue removable insole'),
    'PU insole': ('Insole', 'Anti-fatigue removable insole'),
    'PU foam system, foamed in place': ('Insulation / lining', 'PU foam insulation'),
    'Polyester knit lining': ('Insulation / lining', 'Knit lining fabric'),
    'Wool-blend felt sheet, liner sewn in-house': ('Insulation / lining', 'Removable felt liner'),
    'Steel toe cap': ('Safety toe', 'Steel toe cap'),
    'Composite toe cap': ('Safety toe', 'Composite toe cap'),
    'Steel midsole plate': ('Puncture plate', 'Steel plate'),
    'Composite aramid plate': ('Puncture plate', 'Composite plate'),
    'Labels, patch, pull loop': ('Trims', 'Labels, patches, pull loops'),
    'Labels, safety patch, pull loop': ('Trims', 'Labels, patches, pull loops'),
    'Retail box and tissue': ('Packaging', 'Retail box and tissue'),
    # Not in her library yet: 'Colour masterbatch', 'Plastic buckles, eyelets, laces'
}


def sync_codes_with_library():
    """Check (and where clear, update) each part's code against Nazia's verified codes."""
    import re
    try:
        import openpyxl
        ws = openpyxl.load_workbook(WORKBOOK, data_only=True)['Verified_Component_Library']
    except Exception as e:
        print(f"Code check SKIPPED (could not read {WORKBOOK.name}: {e}). Codes typed above are used.")
        return
    def as_code(v):                          # Excel may store 6406.10 as the number 6406.1
        return f"{v:.2f}" if isinstance(v, float) else str(v or "")
    lib = {(str(r[0]).strip(), str(r[1]).strip()): as_code(r[9]) for r in ws.iter_rows(min_row=2, values_only=True) if r[0]}
    report = {"updated": [], "ok": 0, "check": [], "not in library": set()}
    for sku, rows in PARTS.items():
        for i, (comp, desc, hs, sup, cost) in enumerate(rows):
            key = LIBRARY_ROW.get(desc)
            if not key or key not in lib:
                report["not in library"].add(f"{desc} ({hs})")
                continue
            theirs = re.findall(r"\d{4}(?:\.\d{2})?", lib[key])
            ours = re.findall(r"\d{4}(?:\.\d{2})?", hs)
            if len(theirs) == 1 and len(theirs[0]) == 7 and ours != theirs:
                rows[i] = (comp, desc, theirs[0], sup, cost)          # her single verified code wins
                report["updated"].append(f"{sku} {desc}: {hs} -> {theirs[0]}")
            elif all(any(o == t or (len(t) == 4 and o.startswith(t)) for t in theirs) for o in ours):
                report["ok"] += 1
            else:
                report["check"].append(f"{sku} {desc}: ours {hs}, library says '{lib[key]}'")
    print(f"Code check against {WORKBOOK.name}: {report['ok']} parts match", end="")
    print(f", {len(report['updated'])} updated from her library" if report["updated"] else "")
    for x in report["updated"]:
        print("   updated:", x)
    for x in report["check"]:
        print("   CHECK:  ", x)
    for x in sorted(report["not in library"]):
        print("   not in her library yet:", x)



def write(name, fields, rows):
    with open(CATALOGUE_FOLDER / name, 'w', newline='', encoding='utf-8') as f:
        w = csv.writer(f)
        w.writerow(fields)
        w.writerows(rows)


def main():
    CATALOGUE_FOLDER.mkdir(exist_ok=True)
    country = {s[0]: s[2] for s in SUPPLIERS}
    problems = [f"{sku}: unknown supplier {p[3]}" for sku, ps in PARTS.items() for p in ps if p[3] not in country]
    problems += [f"{b['sku']}: no parts listed" for b in BOOTS if b['sku'] not in PARTS]
    if problems:
        raise SystemExit("Fix these first:\n  " + "\n  ".join(problems))
    write('suppliers.csv', ['supplier_id', 'name', 'country', 'supplies', 'declaration_status', 'declaration_expiry', 'language'], SUPPLIERS)
    write('boots.csv', BOOT_FIELDS, [[b[k] for k in BOOT_FIELDS] for b in BOOTS])
    write('parts.csv', ['sku', 'line_no', 'component', 'description', 'hs_code', 'supplier_id', 'origin_country', 'cost_cad'],
          [[sku, i, c, d, hs, s, country[s], cost] for sku, ps in PARTS.items() for i, (c, d, hs, s, cost) in enumerate(ps, 1)])
    n_parts = sum(len(v) for v in PARTS.values())
    print(f"Catalogue written to {CATALOGUE_FOLDER}: {len(BOOTS)} boots, {n_parts} parts, {len(SUPPLIERS)} suppliers")
    for b in BOOTS:
        print(f"   {b['sku']:8} {b['style_name'][:30]:30} {len(PARTS[b['sku']]):2} parts   {b['tests'][:60]}")


main()
```
